### Pricing a Forward with an Unknown Strike

Security pays $S(T_2)-S(T_1)$  at time $T_2$ ($0<T_1<T_2$). $S(t)$ is the price of underlying asset at time $t$. $r$ is an annual risk free interest rate.

a.  () Assume security does not pay dividends. What is its currect price (at $t=0$)?

b. () Now assume security pays continuous dividend $q$. What is the current price in this case?

**Part (a)**. I use portfolio replication. The security has the following cash flows, where $P$ is its current price:

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> -P  </font>   | <font size=3>    0 </font> | <font size=3>    S(T_2) - S(T_1) </font> |

I construct a portfolio with zero net cash flow at $T_1$ and a payoff of $S(T_2)-S(T_1)$ at $T_2$. Its initial value must equal the security price. The replication proceeds in three steps:

**Step 1:** Buy one unit of the underlying at $t=0$ for $S(0)$ and sell it at $T_2$ for $S(T_2)$.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> -S(0)  </font>   | <font size=3>    0 </font> | <font size=3>    S(T_2) </font> |

**Step 2:** At $T_1$, borrow $e^{-r(T_2-T_1)}S(T_1)$ until $T_2$ (the spot price is known at $T_1$). The repayment at $T_2$ is $S(T_1)$.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> 0  </font>   | <font size=3>    e^{- r(T_2 - T_1)}*S(T_1)  </font> | <font size=3>    -S(T_1) </font> |

**Step 3:** Offset the cash inflow at $T_1$ by shorting $e^{-r(T_2-T_1)}$ units of the underlying at $t=0$ and closing the short at $T_1$.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> e^{- r(T_2 - T_1)}*S(0)  </font>   | <font size=3>    -e^{- r(T_2 - T_1)}*S(T_1)  </font> | <font size=3>    0 </font> |


All future cash flows are replicated, so the security price is:

$$-P = -S(0) + e^{- r(T_2 - T_1)}*S(0)$$

$$P = S(0) * (1 - e^{- r(T_2 - T_1)})$$

**Part (b)**. With a continuous dividend yield $q$, holding $e^{-qT}$ units at $t=0$ and reinvesting dividends produces one unit at $T$: $e^{-qT}\cdot e^{qT}=1$. I adjust the numbers of shares in the replicating portfolio accordingly.

**Step 1:** Buy $e^{-qT_2}$ units of the underlying at $t=0$; with reinvested dividends, this becomes one unit at $T_2$.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> -e^{-qT_2} * S(0)  </font>   | <font size=3>    0 </font> | <font size=3>    S(T_2) </font> |

**Step 2:** The borrowing leg remains unchanged.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> 0  </font>   | <font size=3>    e^{- r(T_2 - T_1)}*S(T_1)  </font> | <font size=3>    -S(T_1) </font> |

**Step 3:** Short $e^{-qT_1}e^{-r(T_2-T_1)}$ units at $t=0$ to offset the cash flow at $T_1$.

|  <font size=4>__t = 0__</font> |  <font size=4>t = T1</font> |  <font size=4>t = T2</font> |
| :--------  | :---: |  :---: |
| <font size=3> e^{-qT_1} * e^{-r(T_2 - T_1)} * S(0)  </font>   | <font size=3>    -e^{- r(T_2 - T_1)}*S(T_1)  </font> | <font size=3>    0 </font> |


The resulting portfolio replicates all cash flows; the current security price is:

$$ -P = -e^{-qT_2} * S(0) + e^{-qT_1} * e^{-r(T_2 - T_1)} * S(0)$$

$$P = S(0) * (e^{-qT_2}  - e^{-qT_1} * e^{-r(T_2 - T_1)} ) $$

### FX Forwards and Mark-to-Market


* The current CHF/EUR exchange rate is $S = 0.9$ (It means one Swiss frank is worth 0.9 euro (or 90 euro cents)).

* The risk free interest rates are $0\%$ (for CHF) and $1\%$ (for EUR) (continuous compounding)

__a)__ () Determine 9 months forward rate for CHF/EUR exchange rate.


__b)__  () The bank enters the forward contract to sell $100m$ CHF to EUR at 9m forward rate (calculated in a)).

 In 3 month the exchange rate changed to $0.95$ , and EUR interest rate goes up to $2\%$.

- Marked to market the forward contract.





__c)__ ()


__Duality__

Mark to market the forward position from b) using alternative risk neutral measure (by switching asset and money). Does the price coinside with what you calculated in b)?

__d)__  ()


__Assume the initial market data for questions, the same as at a) point__

Calculate the fair par forward exchange rate to buy CHF for USD at the following schedule (in one year (1Y) we buy 1 m CHF and so on):

  |1Y |2Y |3Y |
  |---|---|---|
  |1m |2m |3m |


**Part (a)** I use the cost-of-carry formula:
$$F(0, T) = e^{(r-q) T} S (0)$$

```python
S, r, q, T = 0.9, 0.01, 0, 9/12
F = np.exp((r-q)*T)*S
print(f"9 months forward rate for CHF/EUR = {F}")
```

**Part (b)** For the short CHF forward position, I apply:

$$V(t,T) = Ne^{-r(T-t)}(𝐹(0,𝑇) - F(t,T))$$


```python
N = 100
S1, r1, q1, t = 0.95, 0.02, 0, 3/12
Fnew = np.exp((r1-q1)*(T - t))*S1
V = np.exp(-r1*(T - t))*(F - Fnew) * N

print(f"Fnew = {Fnew} M EUR")
print(f"V = {V} M EUR")
```

**Part (c)** I switch the underlying and the numeraire and repeat the valuation. In the inverse quotation, this corresponds to a long EUR position, so the sign changes.

```python
N_2 = 100 * F # express the contract notional in EUR
r2 = 0 # CHF is now the numeraire

V = - np.exp(-r2*(T - t))*(1/F -1/Fnew) * N_2
V_updated = V * S1
print(f"V_CHF = {V} M CHF")
print(f"V_EUR = {V_updated} M EUR")
```

The mark-to-market value agrees with part (b).

**Part (d)** For an exchange rate $K$, the cash flows at each $t_i$ are:

* The company pays the bank $N_i\cdot K$ EUR. Its present value is:

$$PV_{EUR} = \sum_{i = 1}^n e^{- r_{EUR} \cdot t_i} \cdot N_i \cdot K$$

* The company receives $N_i$ CHF from the bank. Its present value is:

$$PV_{CHF} = \sum_{i=1}^n e^{-r_{CHF} \cdot t_i} N_i$$

The fair rate equates the present values of the two legs:

$$S(0) \cdot PV_{CHF} = PV_{EUR}$$

I solve this linear equation for $K$.

```python
r_EUR = 0.01
r_CHF = 0
S = 0.9
N = [1, 2, 3]

PV_EUR_divided_by_K = sum(np.exp(-r_EUR * (i + 1)) * N[i] for i in range(len(N)))
PV_CHF = sum(np.exp(-r_CHF * (i + 1)) * N[i] for i in range(len(N)))

K = PV_CHF * S / PV_EUR_divided_by_K

print(f'K = {K}')
```

### Forward and Futures Valuation


Consider stock with current price $S$ = 100 USD. It pays two dividends of 2 USD in 3m and 2 USD in 9m. Interest rate= 1% (continuous compounding)

a) () What is one-year forward price for this stock?

b) () Assume instead, the market forward price is equal to $\$90$.
    Please, explain in details, how to realize the arbitrage opportunity, and how much money     you make by trading 1 forward contract.

**Part (a)** I calculate the forward price from the spot price net of the present value of dividends:

$$F(0,T) = e^{rT}(S(0)-I_0)$$

```python
r = 0.01
T1 = 3
T2 = 9
T = 12
D = 2
S_0 = 100
I_0 = np.exp(-r*T1/12) * D + np.exp(-r*T2/12) * D

F_0 = np.exp(r*T/12) * (S_0 - I_0)
print(f'One-year forward price F(0,T) = {F_0}')
```

**Part (b)** The market forward is underpriced (90 versus approximately 96.99). I take a long position in the cheap forward and a short position in its synthetic replication. The cash flows are:

0. Cash flows of the long forward:

|  <font size=4>__t = 0__</font> |  <font size=4>t = 0.25 </font> |  <font size=4>t = 0.75 </font> | <font size=4>t = 1</font> |
| :--------  | :---: |  :---: |  :---: |
| <font size=3> 0  </font>   | <font size=3>    0 </font> | <font size=3> 0 </font> | <font size=3>  S(1) - 90 </font> |

1. Deposit $90e^{-r}$ dollars until $t=1$:

|  <font size=4>__t = 0__</font> |  <font size=4>t = 0.25 </font> |  <font size=4>t = 0.75 </font> | <font size=4>t = 1</font> |
| :--------  | :---: |  :---: |  :---: |
| <font size=3> -e^{-r}*90  </font>   | <font size=3>    0 </font> | <font size=3> 0 </font> | <font size=3>  90 </font> |

2. Short the underlying until $t=1$:

|  <font size=4>__t = 0__</font> |  <font size=4>t = 0.25 </font> |  <font size=4>t = 0.75 </font> | <font size=4>t = 1</font> |
| :--------  | :---: |  :---: |  :---: |
| <font size=3> 100  </font>   | <font size=3>    -2 </font> | <font size=3> -2 </font> | <font size=3>  -S(1) </font> |

3. Deposit $2e^{-0.25r}$ until $t=0.25$ and $2e^{-0.75r}$ until $t=0.75$ to cover dividends on the short position:

|  <font size=4>__t = 0__</font> |  <font size=4>t = 0.25 </font> |  <font size=4>t = 0.75 </font> | <font size=4>t = 1</font> |
| :--------  | :---: |  :---: |  :---: |
| <font size=3> -2*e^{-0.25r} - 2*e^{-0.75r}  </font>   | <font size=3>    2 </font> | <font size=3> 2 </font> | <font size=3>  0 </font> |

Adding all cash flows, including the forward, gives:

|  <font size=4>__t = 0__</font> |  <font size=4>t = 0.25 </font> |  <font size=4>t = 0.75 </font> | <font size=4>t = 1</font> |
| :--------  | :---: |  :---: |  :---: |
| <font size=3> 100 - e^{-r}*90 - 2*e^{-0.25r} - 2*e^{-0.75r}  </font>   | <font size=3>    0 </font> | <font size=3> 0 </font> | <font size=3>  0 </font> |

The arbitrage proceeds per contract are $100-90e^{-r}-2e^{-0.25r}-2e^{-0.75r}$. I calculate their value below:



```python
Pr = 100 - np.exp(-r)*90 - np.exp(-r*T1/12) * D - np.exp(-r*T2/12) * D
print(f'Arbitrage proceeds per contract: {Pr} $')
```

### Arbitrage and Replication

Topic: __Futures__

a) () The spot price of gold is currently \\$ 530 per troy ounce. A gold futures contract with maturity in 3 years is currently priced at  \\$614 per ounce. The 3-year risk-free interest rate (expressed per annum with continuous compounding) is equal to 4.5%. What is the implied storage of cost of gold per annum expressed with continuous compounding?


b) () Two futures contracts are traded on a financial asset without dividends: a one-month
contract and a one-year contract. The one-month futures price is $F1= \$ 100$ and the one-
year futures price is $F2=\$105$. Assume interest rates are constant and futures and forward
prices are equal.

What is the spot price of the underlying asset?

**Part (a)** The cost-of-carry formula is:

$$F_0(T) =e^{(r+u)T}S$$, where $u$ are implied storage costs per annum as a proportion of the spot price.

```python
S = 530
T = 3
F_0 = 614
r = 0.045

def equation_to_solve(u):
    left_term = F_0
    right_term = np.exp((r + u) *T) * S
    return left_term - right_term

u_initial_guess = 0.5
u_solution = fsolve(equation_to_solve, u_initial_guess)

print(f"The solution for u is: {u_solution[0]} or {100 * u_solution[0]} %")
```

**Part (b)** With no dividends or storage costs, the cost of carry is the risk-free rate $r$, so:

$$F = e^{rT}S$$

I solve the system:

$$100 = e^{r/12}S$$
$$105 = e^{r}S$$

Dividing the equations gives:

$$\frac{F1}{F2} = e^{r(T1 - T2)}$$

Taking logarithms:

$$log(\frac{F1}{F2}) =r(T1 - T2)$$

$$r = \frac{log(\frac{F1}{F2})}{T1 - T2} $$

Therefore,

$$S = \frac{F2}{e^{\frac{log(\frac{F1}{F2})}{T1 - T2}}}$$

```python
F1 = 100
F2 = 105

T1 = 1/12
T2 = 1

r = np.log(F1 / F2) / (T1 - T2)
S = F2 / np.exp(r)

print(f'Spot price of the underlying asset = {S}')
```
