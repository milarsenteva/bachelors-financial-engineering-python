### Convexity of European Call Prices

() Prove the convexity property of European call prices (see lecture notes).

**Convexity of European call prices:** $$C(K_2) \le \frac{K_3-K_2}{K_3-K_1}C(K_1)+\frac{K_2-K_1}{K_3-K_1}C(K_3),\quad K_1<K_2<K_3.$$

The inequality holds if the payoff on the left never exceeds the payoff of the portfolio on the right. I compare both payoffs state by state:

Left-hand-side payoff: $$Payoff_L = (S_T - K_2)^{+}$$

1. $S_T < K_1 : Payoff_L = 0$

2. $ K_1 <= S_T < K_2 : Payoff_L = 0$

3. $ K_2 <= S_T < K_3 : Payoff_L = S_T - K_2 $

4. $S_T >= K_3 : Payoff_L = S_T - K_2 $

Right-hand-side payoff: $$Payoff_R = \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1)^{+} + \frac{K_2 - K_1}{K_3 - K_1}(S_T - K_3)^{+}$$

1. $S_T < K_1 : Payoff_R = 0$

2. $ K_1 <= S_T < K_2 : Payoff_R = \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) $

3. $ K_2 <= S_T < K_3 : Payoff_R = \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) $

4. $S_T >= K_3 : Payoff_R = \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) + \frac{K_2 - K_1}{K_3 - K_1}(S_T - K_3) $

I compare the payoffs in each possible state:

1. $S_T < K_1 : 0 >= 0$ - **holds**

---

2. $ K_1 <= S_T < K_2 : \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) >= 0$ - **holds**, because in this state $S_T >= K_1$

---

3. $ K_2 <= S_T < K_3 : \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) >= S_T - K_2 $ - to verify this inequality, expand both sides

$$ \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) >= S_T - K_2 $$

$$ K_2 - \frac{K_3 - K_2}{K_3 - K_1} K_1 >= S_T (1 - \frac{K_3 - K_2}{K_3 - K_1}) $$

$$ K_2 (K_3 - K_1) - (K_3 - K_2) K_1 >= S_T (K_3 - K_1 - K_3 + K_2) $$ - multiplication is valid because $K_3 > K_1$ by assumption

$$ K_2 K_3 - K_2 K_1 - K_3 K_1 + K_2 K_1 >= S_T ( K_2 - K_1) $$

$$ K_3 (K_2 - K_1)  >= S_T ( K_2 - K_1) $$  - division is valid because $K_2 > K_1$ by assumption

$$ K_3 >= S_T $$ - **holds** in this state ($S_T < K_3$)

---

4. $S_T >= K_3 : \frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) + \frac{K_2 - K_1}{K_3 - K_1}(S_T - K_3) >=  S_T - K_2$ - to verify this inequality, expand both sides

$$\frac{K_3 - K_2}{K_3 - K_1}(S_T - K_1) + \frac{K_2 - K_1}{K_3 - K_1}(S_T - K_3) >=  S_T - K_2$$ - multiplication is valid because $K_3 > K_1$ by assumption

$$(K_3 - K_2)(S_T - K_1) + (K_2 - K_1)(S_T - K_3) >=  (K_3 - K_1)(S_T - K_2)$$

$$(K_3 - K_2 + K_2 - K_1)S_T - K_3K_1 + K_2K_1 - K_2K_3 + K_1K_3  >=  (K_3 - K_1)S_T - K_2K_3 + K_1K_2$$

$$(K_3 - K_1)S_T + K_2K_1 - K_2K_3  >=  (K_3 - K_1)S_T - K_2K_3 + K_1K_2$$ - both sides are identical, so the inequality **holds**

The left-hand-side payoff is no greater in any state. By no-arbitrage pricing, this proves convexity of European call prices in the strike.

### Option Spread Arbitrage

We observe the following prices of call options ($S = \$100$, $r=5\%$, $T = 1Y$):

  K  |95 |100|105|
  ---|---|---|---|
  C  |18 |13 |10 |



____ a) Is there an arbitrage here? If yes, explain shorty how to realize it.

I check the following call-spread bound for each pair of strikes:

$$C(K_1) - C(K_2) <= e^{-rT}(K_2 - K_1), K_1 < K_2$$

```python
K_1, K_2, K_3 = 95, 100, 105
C_1, C_2, C_3 = 18, 13, 10
r, T = 0.05, 1

print(f'K_1 and K_2: {C_1 - C_2} <= {np.exp(-r*T)* (K_2 - K_1)} - violated')
print(f'K_2 and K_3: {C_2 - C_3} <= {np.exp(-r*T)* (K_3 - K_2)}')
print(f'K_1 and K_3: {C_1 - C_3} <= {np.exp(-r*T)* (K_3 - K_1)}')
```

The call-spread bound is violated for $K_1$ and $K_2$. I sell the call with strike $K_1$ and buy the call with strike $K_2$, receiving $C(K_1)-C(K_2)$ initially. At maturity:

1. If $S_T<K_1$, both calls expire: the net option payoff is $0$.
2. If $K_1\le S_T<K_2$, only the short call is exercised: the payoff is $K_1-S_T\le0$.
3. If $S_T\ge K_2$, both calls are exercised: the payoff is $K_1-K_2$.

The maximum loss at maturity is $K_2-K_1$. Reserving its discounted value leaves a strictly positive initial surplus:

$$C(K_1)-C(K_2)-e^{-rT}(K_2-K_1)=5-4.756>0.$$

To lock in this arbitrage, invest $e^{-rT}(K_2-K_1)$ at time zero. At maturity:

- The deposit pays $K_2-K_1$.
- The short call spread can lose at most $K_2-K_1$.

Thus, the terminal cash flow is nonnegative in all states, while the time-zero proceeds are strictly positive.

### Put–Call Parity and Synthetic Payoffs

____ Consider the following combination of options: buying a call with strike K1, selling a call with strike K2, buying a put with strike K2, and selling a put with strike K1 (K2 > K1). That is, this is a combination of a bull call spread and a bear put spread.

Do you have enough information to find the current price of this product? If yes, please calculate the price.

If not, what additional info do you need?


I use put–call parity to price this combination:

$$\text{Price}=C(K_1)-C(K_2)+P(K_2)-P(K_1)=[C(K_1)-P(K_1)]-[C(K_2)-P(K_2)].$$

The underlying-specific term is the same for both strikes:

1. No dividends: $C-P=S-e^{-rT}K$.
2. Known cash dividends: $C-P=S-I_0-e^{-rT}K$.
3. Continuous dividend yield: $C-P=e^{-qT}S-e^{-rT}K$.

Writing the shared component as $A(S)$ gives $C-P=A(S)-e^{-rT}K$. Therefore,

$$\text{Price}=e^{-rT}(K_2-K_1).$$

A numerical value would also require the interest rate $r$ and the two strikes.

### No-Arbitrage Bounds for FX Options


____
The current EUR/USD exchange rate is S=1.2. The risk free interest rates are 2% (for EUR) and 0% (for USD) (continuous compounding).

Call option with strike $K=1.2$, $T=1Y$ equal to $5$ USD cents. What is the upper and lower limits for call options with $K_1 = 1.1$ and $K_3=1.3$ and $T=1Y$?

I begin with the general lower and upper bounds for calls on a dividend-paying underlying:

$$(e^{-qT}S-e^{-rT}K)^{+}\le C \le e^{-qT}S$$

Hence:

```python
S = 1.2
K1, K2, K3 = 1.1, 1.2, 1.3
r, q, T = 0, 0.02, 1
C2 = 0.05

lower_bound_C1 = max(0, np.exp(-q * T) * S - np.exp(-r * T) *  K1)
upper_bound_C1 = np.exp(-q * T) * S

lower_bound_C3 = max(0, np.exp(-q * T) * S - np.exp(-r * T) *  K3)
upper_bound_C3 = np.exp(-q * T) * S

print(f' {round(lower_bound_C1, 4)} <= C1 <= {round(upper_bound_C1, 4)}')
print(f' {lower_bound_C3} <= C3 <= {round(upper_bound_C3, 4)}')
```

I also use the call-spread bounds:

$$C(K_1) - C(K_2) <= e^{-(r)T}(K_2 - K_1)$$

$$C(K_2) - C(K_3) <= e^{-(r)T}(K_3 - K_2)$$

They imply:

$$C(K_1) <= e^{-(r)T}(K_2 - K_1) + C(K_2) $$

$$ C(K_3) >= C(K_2) - e^{-(r)T}(K_3 - K_2)$$


```python
print(f' C1 <= {round(np.exp(-r * T) * (K2 - K1) + C2, 4)}')
print(f' C3 >= {round( - np.exp(-r * T) * (K3 - K2) + C2, 4)}')
```

Combining the intrinsic bounds with the call-spread bounds gives

$$0.0762\le C(K_1)\le0.15,\qquad 0\le C(K_3)\le0.05.$$

The upper bound for $C(K_3)$ follows from decreasing call prices as the strike increases.

The outer pair of strikes provides an additional restriction:

$$C(K_1) - C(K_3) <= e^{-(r)T}(K_3 - K_1)$$

It follows that:

$$C(K_1) <= C(K_3) + e^{-(r)T}(K_3 - K_1) <= C(K_3)_{upper} + e^{-(r)T}(K_3 - K_1)$$

$$ C(K_3) >= C(K_1) -  e^{-(r)T}(K_3 - K_1) >= C(K_1)_{lower} -  e^{-(r)T}(K_3 - K_1) $$

```python
print(f' C1 <= {round(np.exp(-r * T) * (K3 - K1) + 0.0762, 4)}')
print(f' C3 >= {round( - np.exp(-r * T) * (K3 - K1) + 0.0762, 4)}')
```

Finally, I use the convexity inequality:

$C(K_2) \le \frac{K_3 - K_2}{K_3 - K_1}C(K_1) + \frac{K_2 - K_1}{K_3 - K_1}C(K_3)$, for $K_1 < K_2 < K_3$  

$$0.05 \le \frac{1}{2}C(K_1) + \frac{1}{2}C(K_3)$$  

$$0.1 \le C(K_1) + C(K_3)$$  

**Combined no-arbitrage restrictions:**

$$0.0762\le C(K_1)\le0.15,\qquad 0\le C(K_3)\le0.05,$$

$$C(K_1)+C(K_3)\ge0.10,\qquad C(K_1)\ge0.05\ge C(K_3).$$

These are **joint constraints**, not two independently selectable intervals.

### Binomial Trees: European and American Options


The current USD/RUB exchange rate is S=70. The risk free interest rates are 5% (for RUB) and 0% (for USD) (continuous compounding). Vol=20%. K=70 for all options in this problem

f) () What is the price of option paying $100 if S_T>70 in T=3 months at t=0

**Part (a)** Build 3-period binomial tree for the stock spanning period of T= 9 months (also calculate u,d,p, and probabilities of all nodes).

```python
S_0, r, sigma, K, T = 70, 0.05, 0.2, 70, 9/12
Delta_t = T/3

u = math.e ** (sigma * math.sqrt(Delta_t))
d = math.e ** (- sigma * math.sqrt(Delta_t))
p = (math.e ** (r * Delta_t) - d)/(u-d)
print(f'u = {x(u)}, d = {x(d)}, p = {x(p)}')
```

```python
# Build the stock-price tree

S_00 = S_0
S_11 = S_00 * u
S_10 = S_00 * d

S_22 = S_11 * u
S_21 = S_11 * d #(same as S_10 * u)
S_20 = S_10 * d

S_33 = S_22 * u
S_32 = S_22 * d #(same as S_21 * u)
S_31 = S_21 * d #(same as S_20 * u)
S_30 = S_20 * d

# Display values:
print("Tree for the XYZ stock price, S:")
print(f'{x(S_00)}')
print(f'{x(S_10)}\t{x(S_11)}')
print(f'{x(S_20)}\t{x(S_21)}\t{x(S_22)}')
print(f'{x(S_30)}\t{x(S_31)}\t{x(S_32)}\t{x(S_33)}')
```

```python
# Build the probability tree

p_00 = 1
p_11 = p
p_10 = 1-p

p_22 = p ** 2
p_21 = math.comb(2, 1) * p * (1-p)
p_20 = (1-p) ** 2

p_33 = p ** 3
p_32 = math.comb(3, 2) * (p ** 2) * (1-p)
p_31 = math.comb(3, 1) * p * ((1-p) ** 2)
p_30 = (1-p) ** 3

# Display values:
print("Tree for probabilities:")
print(f'{x(p_00)}')
print(f'{x(p_10)}\t{x(p_11)}')
print(f'{x(p_20)}\t{x(p_21)}\t{x(p_22)}')
print(f'{x(p_30)}\t{x(p_31)}\t{x(p_32)}\t{x(p_33)}')
```

**Part (b)** What is the price of European Call option with maturity T=3m?

```python
T_C = 3./12.
# Calculate terminal payoffs:
C_11 = max(S_11 - K, 0)
C_10 = max(S_10 - K, 0)

print(f'{x(C_10)}\t{x(C_11)}')
```

```python
C_00 = (p_11 * C_11 + p_10 * C_10) * (math.e ** ( - r * T_C))
print("Price of European Call option with maturity T=3m is:  "+x(C_00))
```

**Part (c)** Explain, how to hedge this option at  t=0 , and demonstrate, that the hedge works.

```python
Delta = (C_11 - C_10)/(S_00 * (u - d))
Delta
```

To delta-hedge a short call, I hold $\Delta$ units of the underlying. I verify that the hedged payoffs in the up and down states are equal, $\pi_u=\pi_d$:

```python
Profit_u = Delta * S_11 - C_11
Profit_d = Delta * S_10 - C_10
print("Profit in case stock price went up:   "+str(Profit_u))
print("Profit in case stock price went down:   "+str(Profit_d))
```

**Part (d)** What is the price of European Put option with maturity T=9m?

```python
T_P = 9/12

P_33 = max(K-S_33, 0)
P_32 = max(K-S_32, 0)
P_31 = max(K-S_31, 0)
P_30 = max(K-S_30, 0)

# Display values:
print(f'{x(P_30)}\t{x(P_31)}\t{x(P_32)}\t{x(P_33)}')
```

```python
P_00 = (p_33 * P_33 + p_32 * P_32 + p_31 * P_31 + p_30 * P_30) * (math.e ** ( - r * T_P))
print("Price of European Put option with maturity T=9m is:  "+x(P_00))
```

**Part (f)** What is the price of American Put option with maturity T=9m?

At maturity ($t=9$ months), the American put has the same terminal payoff as the European put. Hence $AP_{3j}=P_{3j}$ for $j\in\{0,1,2,3\}$.

```python
AP_33 = P_33
AP_32 = P_32
AP_31 = P_31
AP_30 = P_30
```

I move backward one time step and compare the risk-neutral **continuation value** with the **immediate exercise value**:

```python
cont_val_22 = math.e ** (-r * Delta_t) * (p * AP_33 + (1-p) * AP_32)
cont_val_21 = math.e ** (-r * Delta_t) * (p * AP_32 + (1-p) * AP_31)
cont_val_20 = math.e ** (-r * Delta_t) * (p * AP_31 + (1-p) * AP_30)
print(f'Cont. values: {x(cont_val_20)}\t{x(cont_val_21)}\t{x(cont_val_22)}')

print("")

print(f'Exer. values: {x(K-S_20)}\t{x(K-S_21)}\t{x(K-S_22)}')
```

At each node, the American put value is the maximum of the continuation and immediate-exercise values (shown below at $t=6$ months):

```python
AP_22 = max(cont_val_22, K-S_22)
AP_21 = max(cont_val_21, K-S_21)
AP_20 = max(cont_val_20, K-S_20)

print(f'{x(AP_20)}\t{x(AP_21)}\t{x(AP_22)}')
```

I repeat the same comparison at $t=3$ months:

```python
cont_val_11 = math.e ** (-r * Delta_t) * (p * AP_22 + (1-p) * AP_21)
cont_val_10 = math.e ** (-r * Delta_t) * (p * AP_21 + (1-p) * AP_20)
print(f'Cont. values: {x(cont_val_10)}\t{x(cont_val_11)}')

print("")

print(f'Exer. values: {x(K-S_10)}\t{x(K-S_11)}')
```

```python
AP_11 = max(cont_val_11, K-S_11)
AP_10 = max(cont_val_10, K-S_10)

print(f'{x(AP_10)}\t{x(AP_11)}')
```

Finally, I calculate the initial American put value:

```python
cont_val_00 = math.e ** (-r * Delta_t) * (p * AP_11 + (1-p) * AP_10)
print(f'Cont. values: {x(cont_val_00)}')

print("")

print(f'Exer. values: {x(K-S_00)}')

print("")

print(f'Price of American Put option with maturity T=9m: {max(cont_val_00, K-S_00)}')
```

**Part (f)** What is the price of option paying $100 if S_T>70 in T=3 months at t=0

The claim pays 100 **USD**, not RUB. It pays only in the up state at three months. Accordingly, the USD-denominated price uses the USD numeraire and the corresponding risk-neutral probability; discounting `100 * p` at the RUB rate would mix currencies.

```python
# The payoff is in USD, while the original tree uses the RUB pricing measure.
# Convert the risk-neutral up probability via the USD numeraire.
T = 3./12
payoff = 100
r_rub, r_usd = 0.05, 0.0
p_usd = p * u * np.exp(-(r_rub - r_usd) * T)
Der = np.exp(-r_usd * T) * p_usd * payoff
print(f"USD-denominated digital payoff price: {Der:.4f} USD")
```
